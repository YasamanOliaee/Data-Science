```python
spotify_data_combined = spotify_data.groupby('track_id').agg({
    'track_name': 'first',  
    'track_artist': 'first',  
    'playlist_genre': lambda x: '/'.join(x.unique()), 
    'playlist_subgenre': lambda x: '/'.join(x.unique()), 
    'danceability': 'first', 
    'energy': 'first',
    'loudness': 'first',  
    'speechiness': 'first', 
    'acousticness': 'first',  
    'instrumentalness': 'first',  
    'liveness': 'first', 
    'valence': 'first',  
    'tempo': 'first',  
    'duration_ms': 'first',  
}).reset_index()


print(spotify_data_combined.head())
```

```python
from sklearn.feature_selection import VarianceThreshold

threshold = 0.01
selector = VarianceThreshold(threshold=threshold)
selector.fit(spotify_data_combined[numeric_columns])

selected_features = [numeric_columns[i] for i in range(len(numeric_columns)) if selector.variances_[i] > threshold]
print(f"Selected Features after Variance Analysis: {selected_features}")
```

```python
from sklearn.model_selection import train_test_split
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, Embedding, Flatten, Concatenate, Dropout, BatchNormalization
import numpy as np

train_data, val_data = train_test_split(scaled_data, test_size=0.2, random_state=42)


numeric_features = spotify_data_combined.select_dtypes(include=['float64', 'int64']).columns
numeric_train = train_data[numeric_features]
numeric_val = val_data[numeric_features]


categorical_features = spotify_data_combined.select_dtypes(include=['object']).columns
categorical_train = {col: train_data[col].astype('category').cat.codes.values for col in categorical_features}
categorical_val = {col: val_data[col].astype('category').cat.codes.values for col in categorical_features}


numeric_input = Input(shape=(len(numeric_features),), name="numeric_input")
numeric_dense = Dense(128, activation='relu')(numeric_input)
numeric_dense = BatchNormalization()(numeric_dense)
numeric_dense = Dense(256, activation='relu')(numeric_dense)
numeric_dense = Dropout(0.3)(numeric_dense)


categorical_inputs = []
categorical_embeddings = []

for feature in categorical_features:
    vocab_size = spotify_data_combined[feature].nunique() + 1
    embed_dim = min(50, vocab_size // 2)
    
    input_layer = Input(shape=(1,), name=f"{feature}_input")
    embedding_layer = Embedding(input_dim=vocab_size, output_dim=embed_dim)(input_layer)
    flatten_layer = Flatten()(embedding_layer)
    
    categorical_inputs.append(input_layer)
    categorical_embeddings.append(flatten_layer)


combined_inputs = Concatenate()([numeric_dense] + categorical_embeddings)


hidden_layer = Dense(512, activation='relu')(combined_inputs)
hidden_layer = BatchNormalization()(hidden_layer)
hidden_layer = Dropout(0.4)(hidden_layer)

hidden_layer = Dense(256, activation='relu')(hidden_layer)
hidden_layer = BatchNormalization()(hidden_layer)
hidden_layer = Dropout(0.4)(hidden_layer)

hidden_layer = Dense(128, activation='relu')(hidden_layer)
hidden_layer = BatchNormalization()(hidden_layer)
hidden_layer = Dropout(0.3)(hidden_layer)

output_layer = Dense(10, activation='softmax', name="cluster_output")(hidden_layer)


model = Model(inputs=[numeric_input] + categorical_inputs, outputs=output_layer)
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
model.summary()


numeric_data_train = numeric_train.values
numeric_data_val = numeric_val.values

x_train = [numeric_data_train] + [np.expand_dims(categorical_train[col], axis=1) for col in categorical_train]
x_val = [numeric_data_val] + [np.expand_dims(categorical_val[col], axis=1) for col in categorical_val]


y_train = np.random.randint(0, 10, size=(numeric_data_train.shape[0],))
y_val = np.random.randint(0, 10, size=(numeric_data_val.shape[0],))


y_train = tf.keras.utils.to_categorical(y_train, num_classes=10)
y_val = tf.keras.utils.to_categorical(y_val, num_classes=10)


history = model.fit(
    x=x_train,
    y=y_train,
    epochs=10,
    batch_size=32,
    validation_data=(x_val, y_val)
)
```

```python

y_pred = model.predict(x_val)


clusters_pred = tf.argmax(y_pred, axis=1).numpy()

from sklearn.metrics import silhouette_score

silhouette = silhouette_score(numeric_data_val, clusters_pred)
print(f"Silhouette Score: {silhouette:.2f}")
```

```python
def recommend_songs_by_name(song_name, spotify_data, model, top_n=2):
    
    input_song = spotify_data[spotify_data['track_name'] == song_name].iloc[0]
    

    input_numeric = np.array(input_song[numeric_features]).reshape(1, -1).astype(np.float32)
    
    
    input_categorical = [
        np.array([spotify_data[feature].astype('category').cat.categories.get_loc(input_song[feature])], dtype=np.float32)
        for feature in categorical_features
    ]

    
    input_pred = model.predict([input_numeric] + input_categorical)
    input_cluster = tf.argmax(input_pred, axis=1).numpy()[0]

   
    all_numeric = spotify_data[numeric_features].values.astype(np.float32)
    all_categorical = [
        np.array(spotify_data[feature].astype('category').cat.codes.values, dtype=np.float32)
        for feature in categorical_features
    ]
    all_preds = model.predict([all_numeric] + all_categorical)
    spotify_data['predicted_cluster'] = tf.argmax(all_preds, axis=1).numpy()

    
    same_cluster_songs = spotify_data[spotify_data['predicted_cluster'] == input_cluster]

    
    distances = np.linalg.norm(same_cluster_songs[numeric_features].values.astype(np.float32) - input_numeric, axis=1)

    
    same_cluster_songs = same_cluster_songs.assign(distance=distances)
    recommended_songs = same_cluster_songs.sort_values(by='distance').head(top_n)

    return recommended_songs[['track_name', 'track_artist']]



song_name = 'Dance Monkey'


recommendations = recommend_songs_by_name(song_name, spotify_data_combined, model)
print(recommendations)
```

```python
from telegram import Update
from telegram.ext import Application, CommandHandler, MessageHandler, filters, ContextTypes
from ytmusicapi import YTMusic
import numpy as np
import tensorflow as tf


TELEGRAM_BOT_TOKEN = 'MY TOKEN BOT FROM BOT FATHER'

application = Application.builder().token(TELEGRAM_BOT_TOKEN).build()
yt = YTMusic()


async def start_command(update: Update, context: ContextTypes.DEFAULT_TYPE):
    await update.message.reply_text("Hi , whats your favorite song?")

async def recommend_songs(update: Update, context: ContextTypes.DEFAULT_TYPE):
    song_name = update.message.text
    
    try:
        recommendations = recommend_songs_by_name(song_name, spotify_data_combined, model)

        response = f"\u2728 similar songs {song_name}:\n"

        for index, row in recommendations.iterrows():
            track_name = row['track_name']
            artist = row['track_artist']
            
            search_query = f"{track_name} {artist}"
            search_results = yt.search(search_query, filter='songs')

            if search_results:
                link = f"https://music.youtube.com/watch?v={search_results[0]['videoId']}"
                response += f"- {track_name} by {artist}: [link]({link})\n"
            else:
                response += f"- {track_name} by {artist}: link not found\n"

        await update.message.reply_text(response, parse_mode='Markdown')
    
    except Exception as e:
        await update.message.reply_text("sorry there is a problem")


application.add_handler(CommandHandler("start", start_command))
application.add_handler(MessageHandler(filters.TEXT & ~filters.COMMAND, recommend_songs))


print("robot is ready...")
application.run_polling()

```
