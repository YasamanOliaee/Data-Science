```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

file_path = r'C:\Users\ASUS\Desktop\spotify_songs.csv'  
spotify_data = pd.read_csv(file_path)


spotify_data.dropna(inplace=True)  
spotify_data.drop(columns=['key'], inplace=True)  

spotify_data = spotify_data.groupby('track_id').agg({
    'track_name': 'first',
    'track_artist': 'first',
    'playlist_genre': lambda x: '/'.join(x.unique()),
    'playlist_subgenre': lambda x: '/'.join(x.unique()),
    'danceability': 'first',
    'energy': 'first',
    'loudness': 'first',
    'speechiness': 'first',
    'acousticness': 'first',
    'instrumentalness': 'first',
    'liveness': 'first',
    'valence': 'first',
    'tempo': 'first',
    'duration_ms': 'first'
}).reset_index()

 
numeric_features = [
    'danceability', 'energy', 'loudness', 'speechiness', 
    'acousticness', 'instrumentalness', 'liveness', 
    'valence', 'tempo', 'duration_ms'
]


scaler = StandardScaler()
scaled_features = scaler.fit_transform(spotify_data[numeric_features])
```

```python
pca = PCA(n_components=5)  

pca_features = pca.fit_transform(scaled_features)
```

```python
kmeans = KMeans(n_clusters=8, random_state=42)

kmeans_labels = kmeans.fit_predict(pca_features)
```

```python
spotify_data['cluster'] = kmeans_labels
```

```python
silhouette_avg = silhouette_score(pca_features, kmeans_labels)

print(f"Silhouette Score: {silhouette_avg}")
```
